### Climbing Stairs


```python
def climbing_stairs_memo(n):
    if n < 0:
        return 0
    memo = {0: 1, 1: 1}

    def count(step):
        if step < 0:
            return 0
        if step not in memo:
            memo[step] = count(step - 1) + count(step - 2)
        return memo[step]

    return count(n)


def climbing_stairs_tabulaion(n):
    if n < 0:
        return 0
    dp = [0] * (n + 1)
    dp[0] = 1
    for step in range(1, n + 1):
        dp[step] = dp[step - 1]
        if step >= 2:
            dp[step] += dp[step - 2]
    return dp[n]


def climbing_stairs_optimized_tabulation(n):
    if n < 0:
        return 0
    if n <= 1:
        return 1
    previous_two, previous_one = 1, 1
    for _ in range(2, n + 1):
        previous_two, previous_one = previous_one, previous_two + previous_one
    return previous_one


# Проверка решений
print("Проверка Climbing Stairs:")
for n in [0, 1, 2, 3, 5, 45]:
    result_memo = climbing_stairs_memo(n)
    result_tabulation = climbing_stairs_tabulaion(n)
    result_optimized = climbing_stairs_optimized_tabulation(n)
    match = result_memo == result_tabulation == result_optimized
    print(f"n={n:2d} | Memo: {result_memo:10d} | Tab: {result_tabulation:10d} | Opt: {result_optimized:10d} | Match: {match}")
```

## Анализ сложности Climbing Stairs
- **Мемоизация:** Время `O(n)`, Память `O(n)` (из-за стека рекурсии и словаря).
- **Табуляция:** Время `O(n)`, Память `O(n)` (массив `dp`).
- **Оптимизированная табуляция:** Время `O(n)`, Память `O(1)` (храним только два предыдущих значения).

*Это оптимальное решение с точки зрения использования памяти.*

# ДЗ для дополнительных баллов (50)


```python
def house_robber_memo(nums):
    memo = {}

    def best(index):
        if index >= len(nums):
            return 0
        if index not in memo:
            memo[index] = max(best(index + 1), nums[index] + best(index + 2))
        return memo[index]

    return best(0)


def house_robber_tabulation(nums):
    if not nums:
        return 0
    dp = [0] * (len(nums) + 1)
    dp[1] = nums[0]
    for count in range(2, len(nums) + 1):
        dp[count] = max(dp[count - 1], nums[count - 1] + dp[count - 2])
    return dp[-1]


def house_robber_optimized_tabulation(nums):
    previous_two, previous_one = 0, 0
    for amount in nums:
        current = max(previous_one, previous_two + amount)
        previous_two, previous_one = previous_one, current
    return previous_one


# Бонус: Дома стоят по кругу
def house_robber_circular(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return max(0, nums[0])

    def best_range(start, end):
        previous_two, previous_one = 0, 0
        for index in range(start, end):
            current = max(previous_one, previous_two + nums[index])
            previous_two, previous_one = previous_one, current
        return previous_one

    # Либо мы грабим первый дом (и не грабим последний), либо наоборот
    return max(best_range(0, len(nums) - 1), best_range(1, len(nums)))


# Проверка решений
print("Проверка House Robber (линейная улица):")
test_cases = [[1, 2, 3, 1], [2, 7, 9, 3, 1], [], [5], [2, 1]]
for values in test_cases:
    res_memo = house_robber_memo(values)
    res_tab = house_robber_tabulation(values)
    res_opt = house_robber_optimized_tabulation(values)
    match = res_memo == res_tab == res_opt
    print(f"{str(values):15} -> Memo: {res_memo:2d} | Tab: {res_tab:2d} | Opt: {res_opt:2d} | Match: {match}")

print("\nПроверка House Robber (круговая улица):")
circular_cases = [[2, 3, 2], [1, 2, 3, 1], [], [5]]
for values in circular_cases:
    print(f"Circular {str(values):15} -> {house_robber_circular(values)}")
```

## Анализ сложности House Robber
- **Мемоизация:** Время `O(n)`, Память `O(n)` (стек рекурсии и кэш).
- **Табуляция:** Время `O(n)`, Память `O(n)` (массив `dp`).
- **Оптимизированная табуляция:** Время `O(n)`, Память `O(1)`.
- **Круговая улица (House Robber II):** Время `O(n)`, Память `O(1)`. Задача сводится к двум линейным задачам: либо мы не грабим первый дом, либо не грабим последний. Мы берем максимум из этих двух вариантов.
