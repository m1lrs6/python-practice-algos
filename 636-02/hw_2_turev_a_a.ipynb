## Задача 1: «Банкомат»




```python
def atm(amount, limits):
    # Номиналы по убыванию для жадного алгоритма
    denominations = [5000, 2000, 1000, 500, 200, 100]
    # Переворачиваем лимиты, чтобы они соответствовали порядку купюр
    limits_rev = limits[::-1]
    ans = [0] * 6

    def backtrack(idx, remaining):
        # Базовый случай: сумма набрана
        if remaining == 0:
            return True
        # Если перебрали все купюры, а сумма не набрана
        if idx == len(denominations):
            return False

        # Берем максимум крупных купюр, но не больше остатка
        max_take = min(limits_rev[idx], remaining // denominations[idx])

        # Пробуем взять от максимума до 0 купюр текущего номинала
        for take in range(max_take, -1, -1):
            ans[idx] = take
            if backtrack(idx + 1, remaining - take * denominations[idx]):
                return True
        return False

    # Обработка краевого случая из ТЗ
    if amount == 0:
        return [0] * 6

    if backtrack(0, amount):
        return ans[::-1] # Возвращаем в исходном порядке (от 100 до 5000)
    return -1
```

## Задача 2: «Подарочная карта»



```python
def gift_card(prices, target):
    seen = {}
    for i, price in enumerate(prices):
        complement = target - price
        # Если дополнение уже встречалось, значит мы нашли пару
        if complement in seen:
            return [seen[complement], i]
        # Записываем текущую цену и её индекс.
        # Делаем это после проверки, чтобы не взять один и тот же товар дважды.
        seen[price] = i
    return []
```

## Проверка решений (Тесты из ТЗ)



```python
print("=== Тестирование Задачи 1 (Банкомат) ===")
print(f"Пример 1: {atm(700, [10, 10, 10, 10, 10, 10])} -> Ожидалось: 2 купюры (например [0, 1, 1, 0, 0, 0])")
print(f"Пример 2: {atm(300, [0, 0, 0, 0, 0, 0])} -> Ожидалось: -1")
print(f"Пример 3: {atm(350, [10, 0, 10, 10, 10, 10])} -> Ожидалось: -1")
print(f"Пример 4: {atm(0, [0, 0, 0, 0, 0, 0])} -> Ожидалось: [0, 0, 0, 0, 0, 0]")
print(f"Стресс-тест (10^9): {atm(1000000000, [1000000]*6)} -> Отработал моментально")

print("\n=== Тестирование Задачи 2 (Подарочная карта) ===")
print(f"Пример 1: {gift_card([2, 7, 11, 15], 9)} -> Ожидалось: [0, 1]")
print(f"Пример 2: {gift_card([3, 3], 6)} -> Ожидалось: [0, 1]")
print(f"Доп. тест с отрицательными числами: {gift_card([-3, 4, -1, 8], 5)} -> Ожидалось: [0, 3]")

# Автоматические проверки (Assertions) для сдачи преподавателю
def check_all():
    # Банкомат
    r = atm(700, [10]*6)
    assert isinstance(r, list) and len(r) == 6
    assert sum(r[i]*[100,200,500,1000,2000,5000][i] for i in range(6)) == 700
    assert sum(r) == 2
    assert atm(300, [0]*6) == -1
    assert atm(350, [10,0,10,10,10,10]) == -1
    assert atm(0, [0]*6) == [0]*6

    # Подарочная карта
    assert sorted(gift_card([2,7,11,15], 9)) == [0,1]
    assert sorted(gift_card([3,3], 6)) == [0,1]
    assert sorted(gift_card([-3, 4, -1, 8], 5)) == [0,3]

    print("\n Все assert-проверки пройдены успешно!")

check_all()
```
